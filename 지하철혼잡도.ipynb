```python
import numpy as np
import pandas as pd

station_order = [
        '오금', '경찰병원', '가락시장', '수서', '일원', '대청', '학여울', '대치', '도곡', '매봉',
        '양재', '남부터미널', '교대', '고속터미널', '잠원', '신사', '압구정', '옥수', '금호',
        '약수', '동대입구', '충무로', '을지로3가', '종로3가', '안국', '경복궁', '독립문',
        '무악재', '홍제', '녹번', '불광', '연신내', '구파발', '지축', '삼송', '원흥', '원당',
        '화정', '대곡', '백석', '마두', '정발산', '주엽', '대화'
    ]

def process_congestion_data(congestion):
    
    # 1. 컬럼명 변경
    congestion = congestion.rename(columns={'5시30분': '6시 이전'})
    
    # 2. 23시 이후 평균 컬럼 생성
    congestion['23시 이후'] = congestion[['23시00분', '23시30분', '00시00분', '00시30분']].mean(axis=1).round(2)
    
    # 3. 불필요 컬럼 제거
    congestion.drop(columns=['23시00분', '23시30분', '00시00분', '00시30분', '역번호', '연번'], inplace=True, errors='ignore')
    
    # 4. 역 순서 및 레이블 매핑
    station_label_map = {station: i for i, station in enumerate(station_order)}
    
    # 5. 정렬 및 결측치 처리
    congestion = congestion.sort_values(by=['요일구분', '상하구분']).reset_index(drop=True)
    congestion = congestion.fillna(0)
    
    # 6. 역번호 컬럼 추가
    congestion['역번호'] = congestion['출발역'].map(station_label_map)
    
    return congestion
```

```python
# 혼잡도 데이터 합치기
def average_congestion_sum_then_divide(congest_2024_1, congest_2024_2, congest_2024_3, congest_2024_4):
    non_time_cols = ['요일구분', '호선', '출발역', '상하구분', '역번호']
    time_cols = [col for col in congest_2024_1.columns if col not in non_time_cols]

    # 외부 조인으로 병합
    merged_df = pd.merge(congest_2024_1, congest_2024_2, on=non_time_cols, how='outer', suffixes=('_1', '_2'))
    merged_df = pd.merge(merged_df, congest_2024_3, on=non_time_cols, how='outer')
    merged_df = pd.merge(merged_df, congest_2024_4, on=non_time_cols, how='outer', suffixes=('_3', '_4'))
    
    # 결측치 처리
    merged_df.fillna(0, inplace=True)

    # 컬럼명 정규화
    for col in time_cols:
        merged_df.rename(columns={col: f'{col}_3'}, inplace=True)
        merged_df.rename(columns={col: f'{col}_4'}, inplace=True)

    # 평균 계산
    result_df = merged_df[non_time_cols].copy()
    for col in time_cols:
        cols_to_sum = [f'{col}_1', f'{col}_2', f'{col}_3', f'{col}_4']
        result_df[col] = merged_df[cols_to_sum].sum(axis=1) / 4

    return result_df

mean_df = average_congestion_sum_then_divide(congestion_2024_1, congestion_2024_2, congestion_2024_3, congestion_2024_4)
mean_df.head()
```

# Step 2. 승하차 데이터 전처리

```python
import numpy as np
import pandas as pd
from datetime import datetime
import holidays
import re

station_order = [
    '오금', '경찰병원', '가락시장', '수서', '일원', '대청', '학여울', '대치', '도곡', '매봉',
    '양재', '남부터미널', '교대', '고속터미널', '잠원', '신사', '압구정', '옥수', '금호',
    '약수', '동대입구', '충무로', '을지로3가', '종로3가', '안국', '경복궁', '독립문',
    '무악재', '홍제', '녹번', '불광', '연신내', '구파발', '지축', '삼송', '원흥', '원당',
    '화정', '대곡', '백석', '마두', '정발산', '주엽', '대화'
]

kr_holidays = holidays.KR(years=[2024])

def get_day_type(date):
    if date in kr_holidays or date.weekday() == 6:
        return '공휴일'
    elif date.weekday() == 5:
        return '토요일'
    else:
        return '평일'
```

```python
def process_subway_data(file_path, station_order):
    
    # 1. 데이터 로드 및 필터링
    df = pd.read_csv(file_path, encoding='cp949')
    df_3 = df[df['호선'] == '3호선'].copy()
    
    # 2. 역명 정제
    df_3['역명'] = df_3['역명'].str.replace(r'\(.*\)', '', regex=True).str.strip()
    
    # 3. 역 순서 라벨링
    station_label_map = {station: i for i, station in enumerate(station_order)}
    df_3['label'] = df_3['역명'].map(station_label_map)
    
    # 4. 정렬
    df_3 = df_3.sort_values(by=['label', '날짜', '구분']).reset_index(drop=True)
    
    # 5. 날짜 처리
    df_3['날짜'] = pd.to_datetime(df_3['날짜'])
    df_3['요일구분'] = df_3['날짜'].apply(get_day_type)
    
    # 6. 불필요 컬럼 제거
    if '역번호' in df_3.columns:
        df_3.drop(columns='역번호', inplace=True)
        
    return df_3
```

# Step 3. 승차량 - 하차량의 데이터프레임 생성

```python
def analyze_passenger_flow(passenger_df):
    
    # 1. 분기 정보 생성
    passenger_df['분기'] = passenger_df['날짜'].dt.to_period('Q')
    
    # 2. 불필요 컬럼 제거
    if '합 계' in passenger_df.columns:
        passenger_df = passenger_df.drop(columns=['합 계'])
    
    # 3. 시간대 컬럼 추출
    time_columns = [col for col in passenger_df.columns 
                   if col not in ['날짜', '호선', '역명', '구분', '요일구분', '분기']]
    
    # 4. 그룹화 및 집계
    grouped = passenger_df.groupby(['분기', '역명', '구분', '요일구분'])[time_columns].sum().reset_index()
    
    # 5. 승차/하차 데이터 분리
    boarding_df = grouped[grouped['구분'] == '승차'].drop('구분', axis=1)
    alighting_df = grouped[grouped['구분'] == '하차'].drop('구분', axis=1)
    
    # 6. 데이터 병합 및 차이 계산
    merged_diff = pd.merge(
        boarding_df,
        alighting_df,
        on=['분기', '역명', '요일구분'],
        suffixes=('_승차', '_하차')
    )
    
    # 7. 시간대별 차이 계산
    for col in time_columns:
        merged_diff[f'{col}_diff'] = merged_diff[f'{col}_승차'] - merged_diff[f'{col}_하차']
    
    # 8. 결과 데이터프레임 생성
    diff_columns = ['분기', '역명', '요일구분'] + [f'{col}_diff' for col in time_columns]
    result_diff = merged_diff[diff_columns]
    
    # 9. 평균 차이 계산
    time_diff_cols = [col for col in result_diff.columns if col.endswith('_diff')]
    result_diff_avg = result_diff.groupby(['역명', '요일구분'])[time_diff_cols].mean().reset_index()
    
    # 10. 분기별 차이 비교 데이터 생성
    merged_comparison = pd.merge(
        result_diff,
        result_diff_avg,
        on=['역명', '요일구분'],
        suffixes=('', '_avg')
    )
    
    for col in time_diff_cols:
        merged_comparison[f'{col}_차이'] = merged_comparison[col] - merged_comparison[f'{col}_avg']
    
    result_comparison = merged_comparison[
        ['분기', '역명', '요일구분'] + 
        [f'{col}_차이' for col in time_diff_cols]
    ]
    
    return result_diff, result_diff_avg, result_comparison
```

```python
def difference_congestion(cong_2024_1, cong_2024):
    # 공통 기준 컬럼과 시간대 컬럼 정의
    key_cols = ['요일구분', '호선', '출발역', '상하구분', '역번호']
    time_cols = [col for col in cong_2024_1.columns if col not in key_cols]

    # 두 데이터프레임 병합 (기준 컬럼으로)
    merged = pd.merge(cong_2024_1, cong_2024, on=key_cols, suffixes=('_1', '_2'))

    # 결과 데이터프레임 초기화 (기준 컬럼 복사)
    diff_df = merged[key_cols].copy()

    # 시간대별 차이 계산 (cong_2024_1 - cong_2024)
    for col in time_cols:
        diff_df[col] = merged[f'{col}_1'] - merged[f'{col}_2']

    return diff_df
```

```python
time_mapping = {
    '6시 이전': '06시 이전_diff_차이',
    '6시00분': '06시-07시_diff_차이', '6시30분': '06시-07시_diff_차이',
    '7시00분': '07시-08시_diff_차이', '7시30분': '07시-08시_diff_차이',
    '8시00분': '08시-09시_diff_차이', '8시30분': '08시-09시_diff_차이',
    '9시00분': '09시-10시_diff_차이', '9시30분': '09시-10시_diff_차이',
    '10시00분': '10시-11시_diff_차이', '10시30분': '10시-11시_diff_차이',
    '11시00분': '11시-12시_diff_차이', '11시30분': '11시-12시_diff_차이',
    '12시00분': '12시-13시_diff_차이', '12시30분': '12시-13시_diff_차이',
    '13시00분': '13시-14시_diff_차이', '13시30분': '13시-14시_diff_차이',
    '14시00분': '14시-15시_diff_차이', '14시30분': '14시-15시_diff_차이',
    '15시00분': '15시-16시_diff_차이', '15시30분': '15시-16시_diff_차이',
    '16시00분': '16시-17시_diff_차이', '16시30분': '16시-17시_diff_차이',
    '17시00분': '17시-18시_diff_차이', '17시30분': '17시-18시_diff_차이',
    '18시00분': '18시-19시_diff_차이', '18시30분': '18시-19시_diff_차이',
    '19시00분': '19시-20시_diff_차이', '19시30분': '19시-20시_diff_차이',
    '20시00분': '20시-21시_diff_차이', '20시30분': '20시-21시_diff_차이',
    '21시00분': '21시-22시_diff_차이', '21시30분': '21시-22시_diff_차이',
    '22시00분': '22시-23시_diff_차이', '22시30분': '22시-23시_diff_차이',
    '23시 이후': '23시-24시_diff_차이'
}
```

```python
# alpha가 1개인 코드

def calculate_alpha_least_squares(df_comparison, cong_1_diff, cong_2_diff, cong_3_diff, cong_4_diff):
    
    # 1. 각 분기 데이터 정리 및 병합 준비
    cong_diffs = {
        '2024Q1': cong_1_diff,
        '2024Q2': cong_2_diff,
        '2024Q3': cong_3_diff, 
        '2024Q4': cong_4_diff
    }
    
    # 2. 데이터 저장용 리스트
    X_q_all = []
    Y_q_all = []
    
    # 3. 각 분기별로 처리
    for quarter, cong_diff in cong_diffs.items():
        # 분기별 Xq 데이터 필터링
        xq_df = df_comparison[df_comparison['분기'] == quarter]
        
        # 매핑 기준 설정 (역명→출발역, 요일구분은 공통)
        xq_keys = ['역명', '요일구분']
        yq_keys = ['출발역', '요일구분']
        
        # xq_df와 cong_diff(yq) 병합
        merged = pd.merge(
            xq_df, 
            cong_diff,
            left_on=xq_keys,
            right_on=yq_keys,
            suffixes=('_xq', '_yq')
        )
        
        # 각 혼잡도 시간대 컬럼에 대해
        for yq_col, xq_col in time_mapping.items():
            if yq_col in cong_diff.columns and xq_col in xq_df.columns:
                # 유효한 데이터만 추출
                valid_rows = merged[~merged[yq_col].isna() & ~merged[xq_col].isna()]
                
                # X_q, Y_q 값 추가
                X_q_all.extend(valid_rows[xq_col].values)
                Y_q_all.extend(valid_rows[yq_col].values)
    
    # 4. 배열로 변환
    X_q = np.array(X_q_all)
    Y_q = np.array(Y_q_all)
    
    # 5. 최소제곱법 수식으로 알파 계산
    alpha = np.sum(X_q * Y_q) / np.sum(X_q ** 2)
    
    return alpha, X_q, Y_q
```

```python
# 시간대별로 alpha를 구하는 코드

def calculate_alpha_per_time(df_comparison, cong_1_diff, cong_2_diff, cong_3_diff, cong_4_diff):
    
    # 1. 각 분기 데이터 정리 및 병합 준비
    cong_diffs = {
        '2024Q1': cong_1_diff,
        '2024Q2': cong_2_diff,
        '2024Q3': cong_3_diff,
        '2024Q4': cong_4_diff
    }

    # 2. 시간대별 X, Y 모으기
    alpha_dict = {}
    for yq_col, xq_col in time_mapping.items():
        X_vals = []
        Y_vals = []

        for quarter, cong_diff in cong_diffs.items():
            # 분기별 Xq 필터
            xq_df = df_comparison[df_comparison['분기'] == quarter]
            merged = pd.merge(
                xq_df,
                cong_diff,
                left_on=['역명', '요일구분'],
                right_on=['출발역', '요일구분'],
                suffixes=('_xq', '_yq')
            )

            if yq_col in merged.columns and xq_col in merged.columns:
                valid = merged[[xq_col, yq_col]].dropna()
                X_vals.extend(valid[xq_col].values.flatten())
                Y_vals.extend(valid[yq_col].values.flatten())

        # 3. α 계산 (해당 시간대)
        X = np.array(X_vals)
        Y = np.array(Y_vals)
        denom = np.sum(X**2)
        if denom != 0:
            alpha_dict[yq_col] = np.sum(X * Y) / denom
        else:
            alpha_dict[yq_col] = np.nan  # 또는 0, 혹은 예외 처리

    return alpha_dict
```

```python
# alpha별로 R^2을 구하는 코드

def calculate_r_squared_per_time(alpha_dict, df_comparison, cong_1_diff, cong_2_diff, cong_3_diff, cong_4_diff):
    
    cong_diffs = {
        '2024Q1': cong_1_diff,
        '2024Q2': cong_2_diff,
        '2024Q3': cong_3_diff,
        '2024Q4': cong_4_diff
    }
    
    r_squared_dict = {}

    for yq_col, xq_col in time_mapping.items():
        alpha = alpha_dict.get(yq_col, None)
        if alpha is None or np.isnan(alpha):
            r_squared_dict[yq_col] = np.nan
            continue
        
        X_vals = []
        Y_vals = []
        for quarter, cong_diff in cong_diffs.items():
            xq_df = df_comparison[df_comparison['분기'] == quarter]
            merged = pd.merge(
                xq_df,
                cong_diff,
                left_on=['역명', '요일구분'],
                right_on=['출발역', '요일구분'],
                suffixes=('_xq', '_yq')
            )
            if yq_col in merged.columns and xq_col in merged.columns:
                valid = merged[[xq_col, yq_col]].dropna()
                X_vals.extend(valid[xq_col].values.flatten())
                Y_vals.extend(valid[yq_col].values.flatten())
        
        X = np.array(X_vals)
        Y = np.array(Y_vals)
        
        if len(X) == 0 or len(Y) == 0:
            r_squared_dict[yq_col] = np.nan
            continue
        
        Y_pred = alpha * X
        
        ss_res = np.sum((Y - Y_pred) ** 2)
        ss_tot = np.sum((Y - np.mean(Y)) ** 2)
        
        # R^2 계산 (총변동 대비 설명된 변동)
        r2 = 1 - ss_res / ss_tot if ss_tot != 0 else np.nan
        r_squared_dict[yq_col] = r2
    
    return r_squared_dict
```

```python
def calculate_alpha_and_r2_by_station(df_comparison, cong_1_diff, cong_2_diff, cong_3_diff, cong_4_diff, single_station_list):
    cong_diffs = {
        '2024Q1': cong_1_diff,
        '2024Q2': cong_2_diff,
        '2024Q3': cong_3_diff,
        '2024Q4': cong_4_diff
    }

    # 결과 저장용 딕셔너리
    station_results = {}

    # 역별 처리
    for station in single_station_list:
        X_total = []
        Y_total = []

        # 해당 역에 대한 데이터 수집
        for quarter, cong_diff in cong_diffs.items():
            xq_df = df_comparison[(df_comparison['분기'] == quarter) & 
                                (df_comparison['역명'] == station)]
            
            cong_df = cong_diff[cong_diff['출발역'] == station]

            merged = pd.merge(
                xq_df,
                cong_df,
                left_on=['역명', '요일구분'],
                right_on=['출발역', '요일구분'],
                suffixes=('_xq', '_yq')
            )

            for y_col, x_col in time_mapping.items():
                if y_col in merged.columns and x_col in merged.columns:
                    valid = merged[[x_col, y_col]].dropna()
                    X_total.extend(valid[x_col].values)
                    Y_total.extend(valid[y_col].values)

        # 역별 계산
        X_total = np.array(X_total)
        Y_total = np.array(Y_total)

        if len(X_total) > 0:
            denom = np.sum(X_total ** 2)
            alpha = np.sum(X_total * Y_total) / denom if denom != 0 else np.nan
            
            Y_pred = alpha * X_total
            ss_res = np.sum((Y_total - Y_pred) ** 2)
            ss_tot = np.sum((Y_total - np.mean(Y_total)) ** 2)
            r2 = 1 - (ss_res / ss_tot) if ss_tot != 0 else np.nan
        else:
            alpha, r2 = np.nan, np.nan

        station_results[station] = (alpha, r2)

    return station_results
```

```python
def preprocess_congestion_data(df, line=7, stations=None):
    # 호선 필터링
    filtered_df = df[df['호선'] == line].copy()
    
    # 역명 클렌징(괄호 내용 제거)
    filtered_df['출발역'] = filtered_df['출발역'].str.replace(r'\(.*\)', '', regex=True).str.strip()
    
    # 온수를 0, 이후 역순으로 1씩 증가
    if stations:
        reversed_stations = stations[::-1]  # 리스트 역순
        station_label = {station: idx for idx, station in enumerate(reversed_stations)}
        filtered_df['label'] = filtered_df['출발역'].map(station_label).fillna(-1).astype(int)
    else:
        filtered_df['label'] = -1

    filtered_df = filtered_df.sort_values(
        ['label', '상하구분'], 
        ascending=[True, True]
    ).reset_index(drop=True)
    
    return filtered_df
```

```python
def convert_hourly_congestion(df):
    
    # 시간대 컬럼 추출 및 정렬
    time_columns = sorted([col for col in df.columns if '시' in col], 
                         key=lambda x: int(x.split('시')[0]))
    
    # 1시간 단위 그룹 생성
    hourly_groups = []
    for i in range(0, len(time_columns), 2):
        group = time_columns[i:i+2]
        if len(group) == 2:
            hourly_groups.append(group)
    
    # 새로운 컬럼명 매핑 (06 ~ 07 형식)
    new_columns = [f"{int(group[0].split('시')[0]):02d} ~ {int(group[0].split('시')[0])+1:02d}" 
                  for group in hourly_groups]
    
    # 평균 계산
    aggregated = {}
    for new_col, cols in zip(new_columns, hourly_groups):
        aggregated[new_col] = df[cols].mean(axis=1)
    
    # 최종 데이터프레임 생성
    result_df = pd.DataFrame({
        'label': df['label'],
        '역명': df['출발역'],
        '구분': df['상하구분'],
        '요일구분': df['요일구분']
    })
    
    for col in new_columns:
        result_df[col] = aggregated[col]
    
    return result_df

congestion_2022_hourly = convert_hourly_congestion(congestion_2022)
congestion_2023_hourly = convert_hourly_congestion(congestion_2023)
congestion_2024_hourly = convert_hourly_congestion(congestion_2024)
```

```python
def process_subway_data(df, start_date, end_date, line='7호선'):
    
    # 필터링 조건 적용
    mask = (
        (df['호선'] == line) &
        (df['날짜'] >= start_date) &
        (df['날짜'] <= end_date)
    )
    filtered_df = df[mask].copy()

    filtered_df = filtered_df.drop(columns=['역번호', '06시 이전', '24시 이후'], errors='ignore')

    start_year = pd.to_datetime(start_date).year
    end_year = pd.to_datetime(end_date).year
    kr_holidays = holidays.KR(years=range(start_year, end_year+1))

    # 요일 구분 함수
    def get_day_type(date):
        if date in kr_holidays or date.weekday() == 6:
            return '공휴일'
        elif date.weekday() == 5:
            return '토요일'
        else:
            return '평일'
    # 요일 구분 컬럼 추가
    filtered_df['요일구분'] = filtered_df['날짜'].apply(get_day_type)
    
    # 역명 클렌징(괄호 내용 제거 및 공백 정리)
    filtered_df['역명'] = filtered_df['역명'].str.replace(r'\(.*\)', '', regex=True).str.strip()
    
    # 표준 역 순서 및 레이블 매핑
    station_label = {station:idx for idx, station in enumerate(stations)}
    
    # 레이블 부여 및 정렬
    if stations:
        reversed_stations = stations[::-1]  # 리스트 역순
        station_label = {station: idx for idx, station in enumerate(reversed_stations)}
        filtered_df['label'] = filtered_df['역명'].map(station_label).fillna(-1).astype(int)
    else:
        filtered_df['label'] = -1
    filtered_df = filtered_df.sort_values(
        ['날짜', 'label', '구분'], 
        ascending=[True, True, True]
    ).reset_index(drop=True)
    
    return filtered_df
```

```python
def time_onehot(df, time_cols):

    # 1. melt로 시간대별로 행을 늘림
    id_vars = [col for col in df.columns if col not in time_cols]
    df_melt = df.melt(
        id_vars=id_vars,
        value_vars=time_cols,
        var_name='시간대',
        value_name='인원수'
    )
    
    # 2. 구분(승차/하차)을 컬럼으로 피벗
    df_pivot = df_melt.pivot_table(
        index=[col for col in df_melt.columns if col not in ['구분', '인원수']],
        columns='구분',
        values='인원수',
        aggfunc='sum'  # 중복 시 합계 계산
    ).reset_index()
    
    # 3. 컬럼명 정리
    df_pivot.columns = [col if col == '시간대' else f'{col}' for col in df_pivot.columns]
    
    # 4. 시간대 원핫 인코딩
    df_final = pd.get_dummies(df_pivot, columns=['시간대', '요일구분'])
    
    
    return df_final

inout_22_ohe = time_onehot(inout_22, time_cols)
inout_23_ohe = time_onehot(inout_23, time_cols)
inout_24_ohe = time_onehot(inout_24, time_cols)
inout_24_09_ohe = time_onehot(inout_24_09, time_cols)
inout_24_03_ohe = time_onehot(inout_24_03, time_cols)
inout_24_06_ohe = time_onehot(inout_24_06, time_cols)
```

```python
def time_label(df, time_cols, time_col_name='시간대', value_col_name='인원수', inout_col='구분'):
    
    # 1. melt로 시간대별 행 확장
    id_vars = [col for col in df.columns if col not in time_cols]
    df_melt = df.melt(
        id_vars=id_vars,
        value_vars=time_cols,
        var_name=time_col_name,
        value_name=value_col_name
    )
    
    # 2. 시간대 레이블 인코딩
    time_labels = {name: idx for idx, name in enumerate(sorted(time_cols))}
    df_melt[f'{time_col_name}_label'] = df_melt[time_col_name].map(time_labels)
    
    # 3. 승차/하차를 컬럼으로 피벗 (행 수 절반 감소)
    df_pivot = df_melt.pivot_table(
        index=[col for col in df_melt.columns if col not in [inout_col, value_col_name]],
        columns=inout_col,
        values=value_col_name,
        aggfunc='sum'  # 중복 시 합계 계산
    ).reset_index()
    
    # 4. 컬럼명 정리
    df_pivot.columns = [
        col if col == time_col_name or col == f'{time_col_name}_label' 
        else f'{col}' 
        for col in df_pivot.columns
    ]
    df_final = pd.get_dummies(df_pivot, columns=['요일구분'])
    
    return df_final

inout_22_label = time_label(inout_22, time_cols)
inout_23_label = time_label(inout_23, time_cols)
inout_24_label = time_label(inout_24, time_cols)
inout_24_09_label = time_label(inout_24_09, time_cols)
inout_24_03_label = time_label(inout_24_03, time_cols)
inout_24_06_label = time_label(inout_24_06, time_cols)
```

```python
def time_cyclic(df, time_cols, time_col_name='시간대', value_col_name='인원수', 
                      inout_col='구분', start_hour=6, end_hour=24, peak_hours=[8, 18]):

    # 1. melt로 시간대별 행 확장
    id_vars = [col for col in df.columns if col not in time_cols]
    df_melt = df.melt(
        id_vars=id_vars,
        value_vars=time_cols,
        var_name=time_col_name,
        value_name=value_col_name
    )
    
    # 2. 시간대에서 시작 시간 추출 (예: '06 ~ 07' → 6)
    df_melt['start_hour'] = df_melt[time_col_name].apply(
        lambda x: int(x.split(' ~ ')[0])
    )
    
    # 3. 주기 인코딩 계산 (사용자 정의 파라미터 적용)
    period = end_hour - start_hour
    shift = peak_hours[0] - (period / 4)  # 예: 8 - 4.5 = 3.5
    angle = 2 * np.pi * (df_melt['start_hour'] - shift) / period
    
    df_melt[f'{time_col_name}_sine'] = np.sin(angle)
    df_melt[f'{time_col_name}_cosine'] = np.cos(angle)
    
    # 4. 불필요한 컬럼 제거
    df_melt = df_melt.drop(columns=['start_hour'])
    
    # 5. 승차/하차를 컬럼으로 피벗 (행 수 절반 감소)
    df_pivot = df_melt.pivot_table(
        index=[col for col in df_melt.columns if col not in [inout_col, value_col_name]],
        columns=inout_col,
        values=value_col_name,
        aggfunc='sum'  # 중복 시 합계 계산
    ).reset_index()
    
    # 6. 컬럼명 정리
    df_pivot.columns = [
        col if col == time_col_name or col in [f'{time_col_name}_sine', f'{time_col_name}_cosine'] 
        else f'{col}' 
        for col in df_pivot.columns
    ]

    df_final = pd.get_dummies(df_pivot, columns=['요일구분'])
    
    return df_final

inout_22_cyclic = time_cyclic(inout_22, time_cols)
inout_23_cyclic = time_cyclic(inout_23, time_cols)
inout_24_cyclic = time_cyclic(inout_24, time_cols)
inout_24_09_cyclic = time_cyclic(inout_24_09, time_cols)
inout_24_03_cyclic = time_cyclic(inout_24_03, time_cols)
inout_24_06_cyclic = time_cyclic(inout_24_06, time_cols)
```

```python
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_percentage_error, mean_absolute_error
feature_cols = ['label', '구분_상선', '구분_하선', '요일구분_공휴일', '요일구분_토요일', '요일구분_평일', 
                '시간대_label', '승차', '하차', '시간대_sine', '시간대_cosine']
label_col = '혼잡도'

X_train = total_22_cyclic[feature_cols]
y_train = total_22_cyclic[label_col]

X_val = total_23_cyclic[feature_cols]
y_val = total_23_cyclic[label_col]

X_test = total_24_cyclic[feature_cols]
y_test = total_24_cyclic[label_col]
```

```python
# 2. Ridge 회귀 최적화
# ------------------------------------------------------------
# 파이프라인 구성
ridge_pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('ridge', Ridge())
])

# 하이퍼파라미터 그리드
ridge_params = {
    'ridge__alpha': [0.001, 0.01, 0.1, 1, 10, 100],
    'ridge__solver': ['svd', 'cholesky', 'lsqr']
}

# 그리드서치 수행
ridge_grid = GridSearchCV(
    ridge_pipe,
    param_grid=ridge_params,
    scoring='neg_mean_squared_error',
    cv=5,
    n_jobs=-1
)
ridge_grid.fit(X_train, y_train)

# 3. XGBoost 최적화
# ------------------------------------------------------------
# 3. XGBoost 최적화 (xgboost.cv 활용)
# ------------------------------------------------------------
import xgboost as xgb
from itertools import product
import numpy as np

# 데이터 DMatrix 변환
dtrain = xgb.DMatrix(X_train, label=y_train)
dval = xgb.DMatrix(X_val, label=y_val)  # 검증 세트

# 하이퍼파라미터 그리드
param_grid = {
    'max_depth': [5, 7],
    'learning_rate': [0.01, 0.05, 0.1],
    'subsample': [0.6, 0.8],
    'colsample_bytree': [0.7, 0.8],
    'gamma': [0, 0.1, 0.2]
}

# 최적 결과 추적용
best_score = np.inf
best_params = None
best_n_estimators = 0

# 그리드 서치 수동 구현
for params in product(*param_grid.values()):
    current_params = dict(zip(param_grid.keys(), params))
    
    # CV 수행
    cv_results = xgb.cv(
        params={**current_params, 'objective': 'reg:squarederror'},
        dtrain=dtrain,
        nfold=3,
        num_boost_round=2000,
        early_stopping_rounds=200,
        metrics='rmse',
        seed=42,
        verbose_eval=False
    )
    
    # 최적 라운드 찾기
    best_round = cv_results.shape[0]
    best_rmse = cv_results['test-rmse-mean'].min()
    
    # 검증 세트로 추가 평가
    model = xgb.train(
        {**current_params, 'objective': 'reg:squarederror'},
        dtrain,
        num_boost_round=best_round,
        evals=[(dval, 'eval')],
        verbose_eval=False
    )
    
    # 성능 비교
    if best_rmse < best_score:
        best_score = best_rmse
        best_params = current_params
        best_n_estimators = best_round

# 최종 모델 훈련
final_xgb = xgb.XGBRegressor(
    **best_params,
    n_estimators=best_n_estimators,
    random_state=42
).fit(X_train, y_train)
```

```python
y_pred_ridge = ridge_grid.best_estimator_.predict(X_test)
y_pred_xgb = final_xgb.predict(X_test)

results = {
    'Ridge Regression': {
        'MSE': mean_squared_error(y_test, y_pred_ridge),
        'R2': r2_score(y_test, y_pred_ridge),
        'MAE': mean_absolute_error(y_test, y_pred_ridge)
    },
    'XGBRegressor': {
        'MSE': mean_squared_error(y_test, y_pred_xgb),
        'R2': r2_score(y_test, y_pred_xgb),
        'MAE': mean_absolute_error(y_test, y_pred_xgb)
    }
}

pd.set_option('display.float_format', '{:.10f}'.format)
# 딕셔너리를 데이터프레임으로 변환
pd.DataFrame(results)
```
